```python
from pyspark.sql.functions import *
```

```python
gold_df = spark.table("default.2_silver")

display(gold_df.limit(10))
```

```python
gold_df.printSchema()
```

```python
from pyspark.sql.functions import sum, round

category_sales = (
    gold_df.groupBy("Category")
    .agg(round(sum("Sales"), 2).alias("Total_Sales"))
    .orderBy(desc("Total_Sales"))
)

display(category_sales)
```

```python
region_sales = (
    gold_df.groupBy("Region")
    .agg(round(sum("Sales"), 2).alias("Total_Sales"))
    .orderBy(desc("Total_Sales"))
)

display(region_sales)
```

```python
product_sales = (
    gold_df.groupBy("Product_Name")
    .agg(round(sum("Sales"), 2).alias("Total_Sales"))
    .orderBy(desc("Total_Sales"))
)

display(product_sales.limit(10))
```

```python
category_profit = (
    gold_df.groupBy("Category")
    .agg(round(sum("Profit"), 2).alias("Total_Profit"))
    .orderBy(desc("Total_Profit"))
)

display(category_profit)
```

```python
# Drop existing streaming table if it exists
spark.sql("DROP TABLE IF EXISTS default.3_gold")

# Save as Delta table
category_sales.write.mode("overwrite").saveAsTable("default.3_gold")
```
