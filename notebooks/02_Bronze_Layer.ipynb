```python
from pyspark.sql.functions import current_timestamp
```

```python
bronze_df = spark.table("workspace.default.bronze_sales")

display(bronze_df.limit(10))
```

```python
bronze_df.printSchema()
```

```python
print("Total Rows :", bronze_df.count())
```

```python
duplicate_rows = (
    bronze_df
    .groupBy("Row_ID")
    .count()
    .filter("count > 1")
)

display(duplicate_rows)
```

```python
from pyspark.sql.functions import col, sum

null_df = bronze_df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in bronze_df.columns
])

display(null_df)
```

```python
bronze_df = bronze_df.withColumn(
    "Ingestion_Time",
    current_timestamp()
)
```

```python
display(bronze_df.limit(10))
```

```python
bronze_df.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("workspace.default.bronze_sales_processed")
```
